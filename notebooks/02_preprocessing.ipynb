# 02 — Preprocessing and quality checks
Inspect crop, resize/pad, and configurable quality measurements.

```python
import pandas as pd
import matplotlib.pyplot as plt
from signature_verification.preprocessing import preprocess_signature
metadata = pd.read_csv('../data/splits/cedar_metadata.csv')
row = metadata.iloc[0]
tensor, quality, processed = preprocess_signature(row.image_path)
print(quality)
plt.imshow(processed, cmap='gray'); plt.axis('off');
```

```python
samples = metadata.groupby('sample_type').sample(4, random_state=42)
fig, axes = plt.subplots(2, 4, figsize=(14, 5))
for axis, (_, sample) in zip(axes.ravel(), samples.iterrows()):
    _, result, image = preprocess_signature(sample.image_path)
    axis.imshow(image, cmap='gray'); axis.set_title(f'{sample.sample_type} | {result.usable}'); axis.axis('off')
plt.tight_layout()
```
